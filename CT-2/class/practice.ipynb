```python
x = np.linspace(-5, 5, 100)
y = x**2

plt.plot(x, y, color='blue', label='x²' )
plt.title("Quadratic Function")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
x = np.linspace(-5,5,100)
data = np.random.rand(500)

# 1 row, 2 columns of plots
fig, axs = plt.subplots(1, 2, figsize=(10, 4))

# Left: line plot
axs[0].plot(x, x**2)
axs[0].set_title("Quadratic")
axs[0].set_xlabel("x")

# Right: histogram
axs[1].hist(data, bins=20, color='orange')
axs[1].set_title("Histogram")

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load built-in datasets
import matplotlib.pyplot as plt

# Load built-in datasets
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")

tips.columns
# total_bill, tip, sex, smoker,
# day, time, size

iris.columns
# sepal_length, sepal_width,
# petal_length, petal_width,
# species
iris = sns.load_dataset("iris")

tips.columns
# total_bill, tip, sex, smoker,
# day, time, size

iris.columns
# sepal_length, sepal_width,
# petal_length, petal_width,
# species

sns.histplot(tips["total_bill"],
             kde=True)
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()

# Scatter plot coloured by category
sns.scatterplot(
  data=tips,
  x="total_bill",
  y="tip",
  hue="sex"      # color by sex
)
plt.title("Tip vs Total Bill")
plt.show()
```

```python
# Step 1: compute correlation matrix
corr = iris.corr(numeric_only=True)

# Step 2: draw heatmap
sns.heatmap(
  corr,
  annot=True,      # show numbers
  cmap='coolwarm'  # blue-red colors
)
plt.title("Correlation Heatmap")
plt.show()

# Other cmap options:
# 'viridis' 'Blues' 'YlOrRd' 'RdBu'
```
