```python
import random
import math

random.seed(1)

#classification

iris_dataset = []

file = open("iris.data", "r")

next(file)

for line in file:

    line = line.strip()

    if line != "":

        parts = line.split(",")

        sample = [
            float(parts[0]),
            float(parts[1]),
            float(parts[2]),
            float(parts[3]),
            parts[4]
        ]

        iris_dataset.append(sample)

file.close()

# split dataset

Train_set = []
Val_set = []
Test_set = []

for sample in iris_dataset:

    r = random.random()

    if r <= 0.7:
        Train_set.append(sample)

    elif r <= 0.85:
        Val_set.append(sample)

    else:
        Test_set.append(sample)

# distance function

def euclidean_distance(A, B):

    total = 0

    for i in range(4):

        total = total + (A[i] - B[i]) ** 2

    return math.sqrt(total)

# prediction

def predict_class(V, K):

    distances = []

    for train_sample in Train_set:

        d = euclidean_distance(V, train_sample)

        distances.append([d, train_sample])

    distances.sort()

    neighbors = distances[:K]

    setosa = 0
    versicolor = 0
    virginica = 0

    for item in neighbors:

        flower = item[1][4]

        if flower == "Iris-setosa":
            setosa += 1

        elif flower == "Iris-versicolor":
            versicolor += 1

        else:
            virginica += 1

    if setosa >= versicolor and setosa >= virginica:
        return "Iris-setosa"

    elif versicolor >= setosa and versicolor >= virginica:
        return "Iris-versicolor"

    else:
        return "Iris-virginica"

# validation

K_values = [1, 3, 5, 10, 15]

results = []

for K in K_values:

    correct = 0

    for sample in Val_set:

        predicted = predict_class(sample, K)

        actual = sample[4]

        if predicted == actual:
            correct += 1

    accuracy = (correct / len(Val_set)) * 100

    results.append([K, accuracy])

# print result

print("\nIRIS DATASET - KNN CLASSIFICATION")
print("__________________________________")
print("K\tValidation Accuracy")
print("___________________________________\n")

for result in results:

    print(result[0], "\t", result[1])

# best K

best_K = results[0][0]
best_accuracy = results[0][1]

for result in results:

    if result[1] > best_accuracy:

        best_accuracy = result[1]
        best_K = result[0]

print("\nBest K =", best_K)

# test accuracy

correct = 0

for sample in Test_set:

    predicted = predict_class(sample, best_K)

    actual = sample[4]

    if predicted == actual:
        correct += 1

test_accuracy = (correct / len(Test_set)) * 100

print("Test Accuracy =", test_accuracy)

#regression

diabetes_dataset = []

file = open("diabetes.csv", "r")

next(file)

for line in file:

    line = line.strip()

    if line != "":

        parts = line.split(",")

        sample = []

        for value in parts:

            sample.append(float(value))

        diabetes_dataset.append(sample)

file.close()

# split dataset

Train_set = []
Val_set = []
Test_set = []

for sample in diabetes_dataset:

    r = random.random()

    if r <= 0.7:
        Train_set.append(sample)

    elif r <= 0.85:
        Val_set.append(sample)

    else:
        Test_set.append(sample)

# distance

def diabetes_distance(A, B):

    total = 0

    for i in range(len(A) - 1):

        total = total + (A[i] - B[i]) ** 2

    return math.sqrt(total)

# prediction

def predict_output(V, K):

    distances = []

    for train_sample in Train_set:

        d = diabetes_distance(V, train_sample)

        distances.append([d, train_sample])

    distances.sort()

    neighbors = distances[:K]

    total = 0

    for item in neighbors:

        total = total + item[1][-1]

    return total / K

# validation

results = []

for K in K_values:

    total_error = 0

    for sample in Val_set:

        predicted = predict_output(sample, K)

        actual = sample[-1]

        total_error = total_error + (actual - predicted) ** 2

    mse = total_error / len(Val_set)

    results.append([K, mse])

# print result

print("\nDIABETES DATASET - KNN REGRESSION")
print("__________________________________")
print("K\tMean Squared Error")
print("___________________________________\n")

for result in results:

    print(result[0], "\t", result[1])

# best K

best_K = results[0][0]
best_MSE = results[0][1]

for result in results:

    if result[1] < best_MSE:

        best_MSE = result[1]
        best_K = result[0]

print("\nBest K =", best_K)

# test mse

total_error = 0

for sample in Test_set:

    predicted = predict_output(sample, best_K)

    actual = sample[-1]

    total_error = total_error + (actual - predicted) ** 2

test_MSE = total_error / len(Test_set)

print("Test Mean Squared Error =", test_MSE)
```
